```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = np.zeros((200, 200), dtype=np.uint8)
cv2.rectangle(img, (40, 40), (160, 160), 255, -1)

plt.imshow(img, cmap='gray')
plt.show()
```

```python
edges = cv2.Canny(img, 50, 150)

plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.show()
```

```python
edges = cv2.Canny(img, 50, 150)

plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title("Detected Edges")
plt.axis('off')

plt.show()
```

```python
edges1 = cv2.Canny(img, 30, 100)
edges2 = cv2.Canny(img, 100, 200)
edges3 = cv2.Canny(img, 150, 250)

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(edges1, cmap='gray')
plt.title("Low Threshold")
plt.axis('off')
```

```python
import urllib.request

url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/fruits.jpg"
urllib.request.urlretrieve(url, "fruit.jpg")

photo = cv2.imread("fruit.jpg")
photo = cv2.cvtColor(photo, cv2.COLOR_BGR2RGB)

gray_photo = cv2.cvtColor(photo, cv2.COLOR_RGB2GRAY)
photo_edges = cv2.Canny(gray_photo, 100, 200)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(photo)
plt.title("Original Photo")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(photo_edges, cmap="gray")
plt.title("Detected Edges")
plt.axis("off")

plt.show()
```
